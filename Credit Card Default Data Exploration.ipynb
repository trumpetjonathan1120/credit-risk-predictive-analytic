```python
#Load the data
df = spark.table("CCC")
display(df.limit(10))

#Check dimensions
print("Rows:", df.count())
print("Columns:", len(df.columns))
print(df.columns)

#Check schema
df.printSchema()

#Check missing values
from pyspark.sql import functions as F

missing = df.select([
    F.sum(F.col(c).isNull().cast("int")).alias(c)
    for c in df.columns
])

display(missing)

#Check target distribution
total = df.count()

default_summary = (
    df.groupBy("dpnm")
      .count()
      .withColumn(
          "percentage",
          F.round(F.col("count") / total * 100, 2)
      )
      .orderBy("dpnm")
)

display(default_summary)


#Basic stats
display(
    df.select(
        "LIMIT_BAL",
        "AGE",
        "BILL_AMT1",
        "PAY_AMT1"
    ).describe()
)

#Check categorical vars
for variable in ["SEX", "EDUCATION", "MARRIAGE"]:
    print(f"\n{variable}")
    df.groupBy(variable).count().orderBy(variable).show()
```
